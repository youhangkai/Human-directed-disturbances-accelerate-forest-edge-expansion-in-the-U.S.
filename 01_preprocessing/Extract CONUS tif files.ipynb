```python
import geopandas as gpd
import rasterio
from shapely.geometry import box
from glob import glob
from tqdm import tqdm
import os
import shutil  # 导入shutil模块

# 加载CONUS范围的shapefile
conus_shapefile_path = 'G:/Hangkai/CONUS Forest Edge Mapping/CONUS shapefile/conus_ard_grid.shp'
conus_boundary = gpd.read_file(conus_shapefile_path)

# 获取2001年地图覆盖文件夹中所有北美(NA)区域的tif文件
tif_directory = 'G:/Public/Glance30/2001_land_cover'
tif_files = glob(os.path.join(tif_directory, '*_NA_*.tif'))

# 检查每个tif文件是否在CONUS范围内
inside_conus_list = []
for tif_file in tqdm(tif_files):
    with rasterio.open(tif_file) as src:
        # 将tif文件的边界转换为Polygon，并确保坐标参考系统匹配
        tif_boundary_geom = box(*src.bounds)
        tif_boundary = gpd.GeoDataFrame(geometry=[tif_boundary_geom], crs=src.crs)
        tif_boundary = tif_boundary.to_crs(conus_boundary.crs)  # CRS转换

        # 检查tif文件的边界是否与CONUS边界相交
        if tif_boundary.intersects(conus_boundary.unary_union).any():
            inside_conus_list.append(tif_file)

# 输出在CONUS范围内的北美NA区域tif文件名
print("North America (NA) TIF files inside CONUS boundary:")
for tif in inside_conus_list:
    print(os.path.basename(tif))

# 定义目标文件夹路径
destination_directory = 'G:/Hangkai/CONUS Forest Edge Mapping/CONUS Landcover/2001'

# 复制每个在CONUS范围内的tif文件到指定目录
for tif_file in tqdm(inside_conus_list):
    shutil.copy(tif_file, destination_directory)

print(f'Copied {len(inside_conus_list)} files to {destination_directory}')
```
