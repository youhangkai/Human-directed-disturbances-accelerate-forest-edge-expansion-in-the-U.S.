## Generate Binary Forest Extent Map from Version 2 Global Land Cover and Land Use 2000 and 2020 Data

```python
import os
import rasterio
from rasterio import mask
from tqdm import tqdm

def create_forest_extent_maps(input_dir, output_dir):
    # Create the output directory if it does not exist
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    # Loop through all TIFF files in the input directory
    for filename in tqdm(os.listdir(input_dir)):
        if filename.endswith('.tif'):
            file_path = os.path.join(input_dir, filename)
            
            # Open the TIFF file
            with rasterio.open(file_path) as src:
                data = src.read(1)  # Read the first band

                # Define forest pixel values
                forest_values = list(range(27, 49)) + list(range(127, 149))
                
                # Create a binary map where forest pixels are 1, others are 0
                forest_map = ((data >= 27) & (data <= 48)) | ((data >= 127) & (data <= 148))
                
                # Check if there are any forest pixels
                if forest_map.sum() == 0:
                    print(f"No forest pixels found in {filename}, skipping.")
                    continue

                # Create a new dataset for output
                output_path = os.path.join(output_dir, filename)
                with rasterio.open(
                    output_path,
                    'w',
                    driver='GTiff',
                    height=src.shape[0],
                    width=src.shape[1],
                    count=1,
                    dtype=rasterio.uint8,
                    crs=src.crs,
                    transform=src.transform
                ) as dst:
                    dst.write(forest_map.astype(rasterio.uint8) * 1, 1)  # Write the binary map to the new TIFF

# Specify the input and output directories
input_dir = r'G:\Public\Global_Land_Cover_and_Land_Use_Change\2005'
output_dir = r'G:\Public\Global_Land_Cover_and_Land_Use_Change\Forest_extent\2005'

# Run the function
create_forest_extent_maps(input_dir, output_dir)
```
