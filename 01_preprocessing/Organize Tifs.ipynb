```python
import os
import shutil
from tqdm import tqdm

# Base directory containing the data
base_dir = "G:\\Hangkai\\CONUS_Forest_Edge_LCMAP\\V08PostReleaseAgent"

# Target directory where the year folders will be created
target_dir = base_dir

def organize_files_by_year(base_dir, target_dir):
    for root, dirs, files in os.walk(base_dir):
        for file in tqdm(files):
            if file.endswith(".tif"):
                # Extract the year from the filename
                year = file.split('_')[2]
                # Create the year directory if it doesn't exist
                year_dir = os.path.join(target_dir, year)
                if not os.path.exists(year_dir):
                    os.makedirs(year_dir)
                # Construct the full file path
                file_path = os.path.join(root, file)
                # Construct the destination path
                destination_path = os.path.join(year_dir, file)
                # Move the file
                shutil.move(file_path, destination_path)
                # print(f'Moved {file} to {year_dir}')

organize_files_by_year(base_dir, target_dir)
```
