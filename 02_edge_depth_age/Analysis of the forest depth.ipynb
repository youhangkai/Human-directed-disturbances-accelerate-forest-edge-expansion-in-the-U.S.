```python
import pandas as pd
import matplotlib.pyplot as plt

# Read CSV files
file_path = 'G:\\Hangkai\\CONUS_Forest_Edge_LCMAP\\Forest_Depth_Classification\\forest_depth_statistics.csv'
df = pd.read_csv(file_path)

# Calculate total area
depth_columns = ['0-30', '30-60', '60-90', '90-120', '>120']
df['Total'] = df[depth_columns].sum(axis=1)

# Calculation of the proportion of each depth category
for column in depth_columns:
    df[f'{column}_Proportion'] = df[column] / df['Total'] * 100

# Visualizing changes in area per forest depth class over time
plt.figure(figsize=(5, 3))
for column in depth_columns:
    plt.plot(df['Year'], df[column], label=column)
plt.title('Area of Each Forest Depth Category Over Years')
plt.xlabel('Year')
plt.ylabel('Area (km2)')
plt.legend(loc='center left', bbox_to_anchor=(1, 0.5))
plt.grid(True)
plt.show()

# Visualize the proportion of each depth category over time
plt.figure(figsize=(5, 3))
for column in [f'{col}_Proportion' for col in depth_columns]:
    plt.plot(df['Year'], df[column], label=column)
plt.title('Proportion of Each Forest Depth Category Over Years')
plt.xlabel('Year')
plt.ylabel('Proportion (%)')
plt.legend(loc='center left', bbox_to_anchor=(1, 0.5))
plt.grid(True)
plt.show()

# Visualizing changes in area per forest depth class over time
plt.figure(figsize=(5, 3))
plt.plot(df['Year'], df['Total'], label="Total forest area")
plt.title('Area of Each Forest Over Years')
plt.xlabel('Year')
plt.ylabel('Area (km2)')
plt.legend(loc='center left', bbox_to_anchor=(1, 0.5))
plt.grid(True)
plt.show()
```

```python
# Define exterior as areas within 100 meters of the edge
# Assuming '0-30' and '30-60' represent areas within 100 meters of the edge
df['Exterior'] = df['0-30'] + df['30-60'] + df['60-90']

# Define interior as the remaining areas
df['Interior'] =  + df['90-120'] + df['>120']

# Calculate proportions for interior and exterior
df['Exterior_Proportion'] = df['Exterior'] / df['Total'] * 100
df['Interior_Proportion'] = df['Interior'] / df['Total'] * 100


# Calculate annual change rates for total, interior, and exterior forest areas
df['Total_Change_Rate'] = df['Total'].pct_change() * 100
df['Interior_Change_Rate'] = df['Interior'].pct_change() * 100
df['Exterior_Change_Rate'] = df['Exterior'].pct_change() * 100

# Visualize changes in total forest area over time
plt.figure(figsize=(10, 6))
plt.plot(df['Year'], df['Total'], label="Total forest area")
plt.title('Total Forest Area Over Years')
plt.xlabel('Year')
plt.ylabel('Area (km2)')
plt.legend()
plt.grid(True)
plt.show()

# Visualize changes in interior and exterior forest area over time
plt.figure(figsize=(10, 6))
plt.plot(df['Year'], df['Interior'], label="Interior forest area")
plt.plot(df['Year'], df['Exterior'], label="Exterior forest area")
plt.title('Interior and Exterior Forest Area Over Years')
plt.xlabel('Year')
plt.ylabel('Area (km2)')
plt.legend()
plt.grid(True)
plt.show()

# Visualize proportions of interior and exterior forest area over time
plt.figure(figsize=(10, 6))
plt.plot(df['Year'], df['Interior_Proportion'], label="Interior forest proportion")
plt.plot(df['Year'], df['Exterior_Proportion'], label="Exterior forest proportion")
plt.title('Proportion of Interior and Exterior Forest Area Over Years')
plt.xlabel('Year')
plt.ylabel('Proportion (%)')
plt.legend()
plt.grid(True)
plt.show()

# Visualize annual change rates
plt.figure(figsize=(10, 6))
plt.plot(df['Year'], df['Total_Change_Rate'], label="Total forest change rate")
plt.plot(df['Year'], df['Interior_Change_Rate'], label="Interior forest change rate")
plt.plot(df['Year'], df['Exterior_Change_Rate'], label="Exterior forest change rate")
plt.title('Annual Change Rate of Forest Areas')
plt.xlabel('Year')
plt.ylabel('Change Rate (%)')
plt.legend()
plt.grid(True)
plt.show()

```

```python
# Create subplots for the area trends
fig, axes = plt.subplots(5, 1, figsize=(14, 20), sharex=True)
for ax, column in zip(axes, depth_columns):
    ax.plot(df['Year'], df[column], marker='o')
    ax.set_title(f'Area of {column} Over Years')
    ax.set_ylabel('Area (km²)')
    ax.grid(True)
axes[-1].set_xlabel('Year')
plt.tight_layout()
plt.show()

# Create subplots for the proportion trends
fig, axes = plt.subplots(5, 1, figsize=(14, 20), sharex=True)
for ax, column in zip(axes, depth_columns):
    proportion_column = f'{column}_Proportion'
    ax.plot(df['Year'], df[proportion_column], marker='o')
    ax.set_title(f'Proportion of {column} Over Years')
    ax.set_ylabel('Proportion (%)')
    ax.grid(True)
axes[-1].set_xlabel('Year')
plt.tight_layout()
plt.show()
```
