```python
import os
import numpy as np
import rasterio
from tqdm import trange

def upscale_back_to_30m(downsampled_path, forest_path, output_path, block_size=33):
    with rasterio.open(downsampled_path) as src1km, rasterio.open(forest_path) as forest_src:
        downsampled_data = src1km.read(1)

        # Step 1: Upsample using repeat
        upsampled_data = np.repeat(np.repeat(downsampled_data, block_size, axis=0), block_size, axis=1)

        # Step 2: Clip to match forest raster shape
        target_height, target_width = forest_src.height, forest_src.width
        upsampled_data = upsampled_data[:target_height, :target_width]

        # Step 3: Copy profile from forest raster
        profile = forest_src.profile.copy()
        profile.update({
            'dtype': src1km.dtypes[0],
            'count': 1,
            'nodata': 0
        })

        # Step 4: Save output with exactly matching shape, transform, CRS
        with rasterio.open(output_path, 'w', **profile) as dst:
            dst.write(upsampled_data, 1)

        print(f"✅ Saved: {output_path} — shape: {upsampled_data.shape}, CRS: {forest_src.crs}")
```

```python
disturbance_1km_folder = r"G:\Hangkai\CONUS_Forest_Edge_LCMAP\1km_dominant_forest_disturbance"
forest_folder = r"G:\Hangkai\CONUS_Forest_Edge_LCMAP\Forest_Depth_Classification"
output_folder = r"G:\Hangkai\CONUS_Forest_Edge_LCMAP\30m_resampled_disturbance"

os.makedirs(output_folder, exist_ok=True)

for year in trange(1985, 2022):
    d_1km_path = os.path.join(disturbance_1km_folder, f"Disturbance_{year}_1km_dominant.tif")
    forest_path = os.path.join(forest_folder, f"LCMAP_CU_{year}_V13_LCPRI.tif")
    out_path = os.path.join(output_folder, f"Disturbance_{year}_30m_resampled.tif")

    if os.path.exists(d_1km_path) and os.path.exists(forest_path):
        try:
            upscale_back_to_30m(d_1km_path, forest_path, out_path)
        except Exception as e:
            print(f"❌ Failed to resample {year}: {e}")
    else:
        print(f"⚠️ Missing input files for {year}")
```
