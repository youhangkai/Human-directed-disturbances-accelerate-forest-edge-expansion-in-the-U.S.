```python
import pandas as pd

# Load the data from the uploaded CSV file
file_path = 'G:/Hangkai/CONUS_Forest_Edge_LCMAP/results.csv'
forest_data = pd.read_csv(file_path)

# Display the first few rows of the dataset to understand its structure
forest_data.head()
```

```python
import matplotlib.pyplot as plt

# Convert the 'Year' column to a string that represents the start year
forest_data['Start Year'] = forest_data['Year'].apply(lambda x: int(x.split(' to ')[0]))+1

# Group by Start Year and Edge Dynamics to calculate the total length for each group
yearly_data = forest_data.groupby(['Start Year', 'Edge Dynamics'])['Length (meters)'].sum().unstack()

# Plotting the trends
plt.figure(figsize=(14, 8))
for column in yearly_data.columns:
    plt.plot(yearly_data.index, yearly_data[column], marker='o', label=column)

plt.title('Trends in Forest Edge Length by Dynamics from 1985 to 2021')
plt.xlabel('Year')
plt.ylabel('Total Edge Length (meters)')
plt.legend(title='Edge Dynamics')
plt.grid(True)
plt.show()
```

```python
# Plotting the trends separately for each type of edge dynamics
fig, axes = plt.subplots(nrows=len(yearly_data.columns), ncols=1, figsize=(14, 20), sharex=True)

for i, column in enumerate(yearly_data.columns):
    axes[i].plot(yearly_data.index, yearly_data[column], marker='o', label=column, lw = 2, c = 'orange')
    axes[i].set_title(f'Trend for {column}', fontsize=14)
    axes[i].set_ylabel('Edge Length (meters)')
    axes[i].grid(True)

plt.xlabel('Year')
fig.suptitle('Separate Trends in Forest Edge Length by Dynamics from 1985 to 2021', fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.985])  # Adjust layout to make room for the global title
plt.show()
```

```python
# Plotting the trends separately for each type of edge dynamics
fig, axes = plt.subplots(nrows=1, ncols=1, figsize=(6, 3), sharex=True)


axes.plot(yearly_data.index, yearly_data['Edge increase due to forest decrease'], marker='o', label=column, lw = 2, c = 'orange')
axes.set_title(f'Trend for Edge increase due to forest decrease', fontsize=14)
axes.set_ylabel('Edge Length (meters)')
axes.grid(True)

plt.xlabel('Year')
plt.tight_layout(rect=[0, 0, 1, 0.985])  # Adjust layout to make room for the global title
plt.show()
```

```python
# Plotting the trends separately for each type of edge dynamics
fig, axes = plt.subplots(nrows=1, ncols=1, figsize=(6, 3), sharex=True)


axes.plot(yearly_data.index, yearly_data['Edge increase due to forest decrease']-yearly_data['Edge decrease due to forest decrease'], marker='o', label=column, lw = 2, c = 'orange')
axes.set_title(f'Trend for Edge net change due to forest area decrease', fontsize=14)
axes.set_ylabel('Edge Length (meters)')
axes.grid(True)

plt.xlabel('Year')
plt.tight_layout(rect=[0, 0, 1, 0.985])  # Adjust layout to make room for the global title
plt.show()
```
