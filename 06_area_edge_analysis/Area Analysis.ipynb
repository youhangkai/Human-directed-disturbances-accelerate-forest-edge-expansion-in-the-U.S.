```python
import os
import rasterio
import numpy as np
import gc

# Directory containing the TIF files
folder = "/Users/hyou/mnt/Hangkai/Hangkai/CONUS_Forest_Edge_LCMAP/Forest_Depth_Classification"

# Year range
years = range(1985, 2022)

# Initialize result dictionary
forest_area_by_year = {}

# Loop over each year
for year in years:
    print(f"Processing {year}...")
    filename = f"LCMAP_CU_{year}_V13_LCPRI.tif"
    filepath = os.path.join(folder, filename)
    
    if not os.path.exists(filepath):
        print(f"File not found: {filepath}")
        continue

    with rasterio.open(filepath) as src:
        data = src.read(1)

        # Use logical condition instead of np.isin to save memory
        forest_mask = (data >= 1) & (data <= 5)
        del data
        gc.collect()

        forest_pixel_count = np.count_nonzero(forest_mask)

        # Optional cleanup
        del forest_mask
        gc.collect()

        # Convert to km² (30 m * 30 m = 0.0009 km²)
        forest_area_km2 = forest_pixel_count * 0.0009
        forest_area_by_year[year] = forest_area_km2

# Print results
print("\n=== Forest Area by Year (km²) ===")
for year, area in forest_area_by_year.items():
    print(f"{year}: {area:.2f} km²")
```
