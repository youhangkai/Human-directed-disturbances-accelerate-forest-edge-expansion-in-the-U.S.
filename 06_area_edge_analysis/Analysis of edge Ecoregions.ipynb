```python
import matplotlib.pyplot as plt

# Define the edge length for each Edge_cod column
edge_length = {
    'Edge_cod_1': 30,
    'Edge_cod_2': 30,
    'Edge_cod_3': 60,
    'Edge_cod_4': 30,
    'Edge_cod_5': 60,
    'Edge_cod_6': 60,
    'Edge_cod_7': 90,
    'Edge_cod_8': 30,
    'Edge_cod_9': 60,
    'Edge_cod10': 60,
    'Edge_cod11': 90,
    'Edge_cod12': 60,
    'Edge_cod13': 90,
    'Edge_cod14': 90,
    'Edge_cod15': 120
}

# Calculate the total edge length for each year and ecoregion
for edge_cod, length in edge_length.items():
    all_data_df[edge_cod] = all_data_df[edge_cod] * length  # Multiply by corresponding edge length

# Calculate the total edge length for each ecoregion and year
all_data_df['Total_Edge_Length'] = all_data_df[[edge_cod for edge_cod in edge_length.keys()]].sum(axis=1)

# List all unique ecoregions
ecoregions = all_data_df['NA_L1NAME'].unique()
print("Unique Ecoregions:", ecoregions)

# Plot time series for each ecoregion in separate subplots
num_ecoregions = len(ecoregions)
fig, axes = plt.subplots(nrows=num_ecoregions//3 + (1 if num_ecoregions % 3 != 0 else 0), ncols=3, figsize=(18, num_ecoregions * 1.5))

# Flatten the axes array for easy iteration
axes = axes.flatten()

for i, ecoregion in enumerate(ecoregions):
    # Filter the data for the specific ecoregion
    ecoregion_data = all_data_df[all_data_df['NA_L1NAME'] == ecoregion]

    # Plot the time series in the subplot
    axes[i].plot(ecoregion_data['Year'], ecoregion_data['Total_Edge_Length']/1000, marker='o')
    axes[i].set_title(f'Total Edge Length in {ecoregion}')
    axes[i].set_xlabel('Year')
    axes[i].set_ylabel('Total Edge Length (km)')
    axes[i].grid(True)

# Hide any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

# Adjust layout
plt.tight_layout()
plt.show()
```

```python
import os
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt

# Set the file path for the consolidated data
consolidated_file = r'G:\Hangkai\CONUS_Forest_Edge_LCMAP\LCMAP_edges\Ecoregion_Classification\consolidated_eco_stats_1985_2021.csv'

# Read the consolidated data from the CSV file
all_data_df = pd.read_csv(consolidated_file)

# Define the edge length for each Edge_cod column
edge_length = {
    'Edge_cod_1': 30,
    'Edge_cod_2': 30,
    'Edge_cod_3': 60,
    'Edge_cod_4': 30,
    'Edge_cod_5': 60,
    'Edge_cod_6': 60,
    'Edge_cod_7': 90,
    'Edge_cod_8': 30,
    'Edge_cod_9': 60,
    'Edge_cod10': 60,
    'Edge_cod11': 90,
    'Edge_cod12': 60,
    'Edge_cod13': 90,
    'Edge_cod14': 90,
    'Edge_cod15': 120
}

# Define the edge label mapping for directions
edge_direction_mapping = {
    'Edge_cod_1': ['West'],
    'Edge_cod_2': ['East'],
    'Edge_cod_3': ['East', 'West'],
    'Edge_cod_4': ['South'],
    'Edge_cod_5': ['South', 'West'],
    'Edge_cod_6': ['South', 'East'],
    'Edge_cod_7': ['South', 'East', 'West'],
    'Edge_cod_8': ['North'],
    'Edge_cod_9': ['North', 'West'],
    'Edge_cod10': ['North', 'East'],
    'Edge_cod11': ['North', 'East', 'West'],
    'Edge_cod12': ['North', 'South'],
    'Edge_cod13': ['North', 'South', 'West'],
    'Edge_cod14': ['North', 'South', 'East'],
    'Edge_cod15': ['North', 'South', 'East', 'West']
}

# Calculate the edge length for each direction
directions = ['North', 'South', 'East', 'West']
for direction in directions:
    all_data_df[direction] = 0

for edge_cod, length in edge_length.items():
    if edge_cod in all_data_df.columns:
        for direction in edge_direction_mapping[edge_cod]:
            all_data_df[direction] += all_data_df[edge_cod] * length / len(edge_direction_mapping[edge_cod])

# Calculate the total edge length for each ecoregion and year
all_data_df['Total_Edge_Length'] = all_data_df[[col for col in edge_length.keys() if col in all_data_df.columns]].sum(axis=1)

# List all unique ecoregions
ecoregions = all_data_df['NA_L1NAME'].unique()
print("Unique Ecoregions:", ecoregions)

# Plot time series for each ecoregion in separate subplots
num_ecoregions = len(ecoregions)
fig, axes = plt.subplots(dpi = 600, nrows=num_ecoregions//3 + (1 if num_ecoregions % 3 != 0 else 0), ncols=3, figsize=(18, num_ecoregions * 1.5))

# Flatten the axes array for easy iteration
axes = axes.flatten()

for i, ecoregion in enumerate(ecoregions):
    # Filter the data for the specific ecoregion
    ecoregion_data = all_data_df[all_data_df['NA_L1NAME'] == ecoregion]

    # Plot the time series in the subplot
    axes[i].plot(ecoregion_data['Year'], ecoregion_data['North']/1000, label='North', lw=2, color = "red", alpha = 0.6)
    axes[i].plot(ecoregion_data['Year'], ecoregion_data['South']/1000, label='South', lw=2, color = "orange", alpha = 0.6)
    axes[i].plot(ecoregion_data['Year'], ecoregion_data['East']/1000, label='East', lw=2, color = "blue", alpha = 0.6)
    axes[i].plot(ecoregion_data['Year'], ecoregion_data['West']/1000, label='West', lw=2, color = "cyan", alpha = 0.6)
    axes[i].set_title(f'Edge Length in {ecoregion}')
    axes[i].set_xlabel('Year')
    axes[i].set_ylabel('Edge Length (km)')
    axes[i].legend()
    axes[i].grid(True)

# Hide any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

# Adjust layout
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming 'all_data_df' is already loaded with the correct data

# Calculate the edge categories within the DataFrame
all_data_df['1_edge'] = all_data_df['Edge_cod_1'] + all_data_df['Edge_cod_2'] + all_data_df['Edge_cod_4'] + all_data_df['Edge_cod_8']
all_data_df['2_edges'] = all_data_df['Edge_cod_3'] + all_data_df['Edge_cod_5'] + all_data_df['Edge_cod_6'] + all_data_df['Edge_cod_9'] + all_data_df['Edge_cod10'] + all_data_df['Edge_cod12']
all_data_df['3_edges'] = all_data_df['Edge_cod_7'] + all_data_df['Edge_cod11'] + all_data_df['Edge_cod13'] + all_data_df['Edge_cod14']
all_data_df['4_edges'] = all_data_df['Edge_cod15']

# Group data by 'Year' and 'NA_L1NAME' and sum the calculated categories
grouped_data = all_data_df.groupby(['Year', 'NA_L1NAME']).agg({
    '1_edge': 'sum',
    '2_edges': 'sum',
    '3_edges': 'sum',
    '4_edges': 'sum'
}).reset_index() 

# Get unique ecoregions
ecoregions = grouped_data['NA_L1NAME'].unique()

# Plot data for each ecoregion
fig, axes = plt.subplots(dpi = 600, nrows=num_ecoregions//3 + (1 if num_ecoregions % 3 != 0 else 0), ncols=3, figsize=(18, num_ecoregions * 1.5))
axes = axes.flatten()

for i, ecoregion in enumerate(ecoregions):
    ecoregion_data = grouped_data[grouped_data['NA_L1NAME'] == ecoregion]
    axes[i].set_title(f'Edge pixels in {ecoregion}')
    axes[i].plot(ecoregion_data['Year'], ecoregion_data['1_edge'], label=f'1 Edge',lw = 2)
    axes[i].plot(ecoregion_data['Year'], ecoregion_data['2_edges'], label=f'2 Edges',lw = 2)
    axes[i].plot(ecoregion_data['Year'], ecoregion_data['3_edges'], label=f'3 Edges',lw = 2)
    axes[i].plot(ecoregion_data['Year'], ecoregion_data['4_edges'], label=f'4 Edges',lw = 2)
    axes[i].set_xlabel('Year')
    axes[i].set_ylabel('Edge Pixels Number')
    axes[i].legend()

# Hide any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])
    
# Show plot
plt.tight_layout()
plt.show()
```
