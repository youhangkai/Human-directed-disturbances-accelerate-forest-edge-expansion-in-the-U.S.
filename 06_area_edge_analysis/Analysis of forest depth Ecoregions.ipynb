```python
import pandas as pd

# Set the file path for the consolidated data
consolidated_file = r'G:\Hangkai\CONUS_Forest_Edge_LCMAP\Forest_Depth_Classification\Ecoregion_Classification\consolidated_eco_stats_1985_2021.csv'

# Read the consolidated data from the CSV file
all_data_df = pd.read_csv(consolidated_file)

# Check for missing values
print("Missing value check below:")
print(all_data_df.isnull().sum())
print("Missing value checked!")

# Fill or drop missing values if necessary
all_data_df = all_data_df.fillna(0)  # Example: fill missing values with 0

# Convert 'Year' column to integer type if it's not already
all_data_df['Year'] = all_data_df['Year'].astype(int)

all_data_df
```

```python
import matplotlib.pyplot as plt

# Example: Plot the time series for a specific ecoregion and Edge_cod column
# List all unique ecoregions
ecoregions = all_data_df['NA_L1NAME'].unique()
print("Unique Ecoregions:", ecoregions)
```

```python
import matplotlib.pyplot as plt


# List all unique ecoregions
ecoregions = all_data_df['NA_L1NAME'].unique()
print("Unique Ecoregions:", ecoregions)

# Plot time series for each ecoregion in separate subplots
num_ecoregions = len(ecoregions)
fig, axes = plt.subplots(dpi = 400, nrows=num_ecoregions//3 + (1 if num_ecoregions % 3 != 0 else 0), ncols=3, figsize=(18, num_ecoregions * 1.5))

# Flatten the axes array for easy iteration
axes = axes.flatten()

for i, ecoregion in enumerate(ecoregions):
    # Filter the data for the specific ecoregion
    ecoregion_data = all_data_df[all_data_df['NA_L1NAME'] == ecoregion]

    # Plot the time series in the subplot
    axes[i].plot(ecoregion_data['Year'], (ecoregion_data['Edge_Dep_1']+ecoregion_data['Edge_Dep_2']+ecoregion_data['Edge_Dep_3'])*0.03*0.03, marker='o',color="red",label="Exterior Forest")
    axes[i].plot(ecoregion_data['Year'], (ecoregion_data['Edge_Dep_4']+ecoregion_data['Edge_Dep_5'])*0.03*0.03, marker='o', color = "green",label="Interior Forest")
    axes[i].set_title(f'Exterior vs Interior Forest Area in {ecoregion}')
    axes[i].set_xlabel('Year')
    axes[i].set_ylabel('Exterior Forest Area (km2)')
    axes[i].grid(True)
    axes[i].legend(loc='center right')

# Hide any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

# Adjust layout
plt.tight_layout()
plt.show()
```
