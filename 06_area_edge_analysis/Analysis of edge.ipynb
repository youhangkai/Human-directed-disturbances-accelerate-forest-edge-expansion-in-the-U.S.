```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data
df = pd.read_csv('G:\\Hangkai\\CONUS_Forest_Edge_LCMAP\\LCMAP_edges\\forest_edge_statistics.csv')

# Define edge code to length lookup (in meters)
edge_length_lookup = {
     1: 30,  2: 30,  3: 60,
     4: 30,  5: 60,  6: 60,  7: 90,
     8: 30,  9: 60, 10: 60, 11: 90,
    12: 60, 13: 90, 14: 90, 15: 120
}

# Compute total edge length in km
df['total_edge_length_km'] = sum(
    df[str(code)] * length for code, length in edge_length_lookup.items()
) / 1000  # convert from meters to kilometers

# Plotting
plt.figure(figsize=(10, 5))
plt.plot(df['year'], df['total_edge_length_km'], marker='o', linewidth=2)
plt.title('Total Forest Edge Length Over Time (1985–2021)')
plt.xlabel('Year')
plt.ylabel('Total Edge Length (km)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Load the data
data = pd.read_csv('G:\\Hangkai\\CONUS_Forest_Edge_LCMAP\\LCMAP_edges\\forest_edge_statistics.csv')

# Calculate total edge length in kilometers for orientations
data['north_edges_km'] = (data['8'] + data['9'] + data['10'] + data['11'] + data['12'] + data['13'] + data['14'] + data['15']) * 30 / 1000
data['south_edges_km'] = (data['4'] + data['5'] + data['6'] + data['7'] + data['12'] + data['13'] + data['14'] + data['15']) * 30 / 1000
data['east_edges_km'] = (data['2'] + data['3'] + data['6'] + data['7'] + data['10'] + data['11'] + data['14'] + data['15']) * 30 / 1000
data['west_edges_km'] = (data['1'] + data['3'] + data['5'] + data['7'] + data['9'] + data['11'] + data['13'] + data['15']) * 30 / 1000

# Plot Yearly Trends for Edge Orientations
fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(6, 8))

# First plot for individual east and north trends
axes[0].plot(data['year'], data['east_edges_km'], label='East', marker='o')
axes[0].plot(data['year'], data['north_edges_km'], label='North', marker='*')
axes[0].set_title('Yearly Trends of Edge Orientations by Length (km)')
axes[0].set_xlabel('Year')
axes[0].set_ylabel('Total Edge Length (km)')
axes[0].legend()
axes[0].grid(True)

# Second plot for combined trends
axes[1].plot(data['year'], data['east_edges_km'] + data['west_edges_km'], label='East+West', marker='o')
axes[1].plot(data['year'], data['north_edges_km'] + data['south_edges_km'], label='North+South', marker='*')
axes[1].set_title('Combined Yearly Trends of Edge Orientations by Length (km)')
axes[1].set_xlabel('Year')
axes[1].set_ylabel('Total Edge Length (km)')
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()  # This adjusts the plots to make sure that the labels and titles do not overlap
plt.show()
```

```python
# Calculate total edge length in kilometers for complexities
data['1_edge'] = (data['1'] + data['2'] + data['4'] + data['8'])
data['2_edges'] = (data['3'] + data['5'] + data['6'] + data['9'] + data['10'] + data['12']) 
data['3_edges'] = (data['7'] + data['11'] + data['13'] + data['14'])
data['4_edges'] = data['15'] 

# Plot Yearly Trends for Edge Complexities
plt.figure(figsize=(12, 6))
plt.plot(data['year'], data['1_edge'], label='1-Edge', marker='o')
plt.plot(data['year'], data['2_edges'], label='2-Edges', marker='o')
plt.plot(data['year'], data['3_edges'], label='3-Edges', marker='o')
plt.plot(data['year'], data['4_edges'], label='4-Edges', marker='o')
plt.title('Yearly Trends of Edge pixles Counts')
plt.xlabel('Year')
plt.ylabel('Total Count')
plt.legend()
plt.grid(True)
plt.show()

# Calculate the total pixels each year (sum of all complexities)
data['total_pixels'] = data['1_edge'] + data['2_edges'] + data['3_edges'] + data['4_edges']

# Calculate the proportions
data['1_edge_prop'] = data['1_edge'] / data['total_pixels']
data['2_edges_prop'] = data['2_edges'] / data['total_pixels']
data['3_edges_prop'] = data['3_edges'] / data['total_pixels']
data['4_edges_prop'] = data['4_edges'] / data['total_pixels']

# Plot the proportions over time
plt.figure(figsize=(12, 6))
plt.plot(data['year'], data['1_edge_prop'], label='1-Edge Proportion', marker='o')
plt.plot(data['year'], data['2_edges_prop'], label='2-Edges Proportion', marker='o')
plt.plot(data['year'], data['3_edges_prop'], label='3-Edges Proportion', marker='o')
plt.plot(data['year'], data['4_edges_prop'], label='4-Edges Proportion', marker='o')
plt.title('Yearly Trends of Proportions of Edge pixles')
plt.xlabel('Year')
plt.ylabel('Proportion')
plt.legend()
plt.grid(True)
plt.show()
```
